## Boxplot and Variance

```kotlin
val moodByYear = dataFrameOf(
    "Jahr" to df.getColumn("Date").toList().map { it.toString().take(4) },
    "Mood" to df.getColumn("Mood").toList().map { (it as Number).toDouble() }
)

moodByYear.plot {
    boxplot("Jahr", "Mood", whiskerIQRRatio = 0.1) {
        boxes {
            alpha = 0.7
            fillColor(Stat.middle) {
                scale = continuous(Color.GREEN..Color.RED)
            }
        }
        outliers {
            color(Stat.x)
            position = Position.jitter(0.1, 0.0)
        }
    }
    layout {
        title = "Mood-Verteilung 2026 und 2027"
        size = 600 to 350
    }
}
```

## t-Tests: Mittelwertvergleich 2026 und 2027

H₀: Die Mittelwerte sind gleich. Beide Tests sind zweiseitig mit α = 0,05.


```kotlin
@file:DependsOn("org.apache.commons:commons-math3:3.6.1")

import java.time.LocalDate
import java.util.Locale
import org.apache.commons.math3.stat.inference.TTest

val moodDates = df.getColumn("Date").toList().map { LocalDate.parse(it.toString()) }
val moodValues = df.getColumn("Mood").toList().map { (it as Number).toDouble() }
require(moodDates.size == moodValues.size) { "Datum und Mood müssen gleich viele Werte enthalten." }
require(moodValues.all { it.isFinite() }) { "Mood darf keine fehlenden oder unendlichen Werte enthalten." }

fun samplesByDayOfYear(year: Int): Map<Int, Double> {
    val rows = moodDates.zip(moodValues).filter { (date, _) -> date.year == year }
    require(rows.size >= 2) { "Für jedes Jahr werden mindestens zwei Werte benötigt." }
    require(rows.map { it.first.dayOfYear }.distinct().size == rows.size) {
        "Für die Paarung darf jeder Tag im Jahr nur einen Mood-Wert enthalten."
    }
    return rows.associate { (date, mood) -> date.dayOfYear to mood }
}

val samples2026 = samplesByDayOfYear(2026)
val samples2027 = samplesByDayOfYear(2027)
require(samples2026.keys == samples2027.keys) {
    "Für die Paarung müssen in beiden Jahren dieselben Tage im Jahr vorhanden sein."
}
val pairedDays = samples2026.keys.sorted()
val mood2026 = pairedDays.map { samples2026.getValue(it) }.toDoubleArray()
val mood2027 = pairedDays.map { samples2027.getValue(it) }.toDoubleArray()
val significanceLevel = 0.05
val meanDifference = mood2026.average() - mood2027.average()
val tTests = TTest()

val pairedT = tTests.pairedT(mood2026, mood2027)
val pairedP = tTests.pairedTTest(mood2026, mood2027)
val studentT = tTests.homoscedasticT(mood2026, mood2027)
val studentP = tTests.homoscedasticTTest(mood2026, mood2027)

fun formatNumber(value: Double) = String.format(Locale.GERMANY, "%.4f", value)
fun testResult(p: Double) =
    if (p < significanceLevel) "Signifikanter Unterschied (H0 verwerfen)"
    else "Kein signifikanter Unterschied (H0 nicht verwerfen)"

println("Mittelwert 2026: " + formatNumber(mood2026.average()))
println("Mittelwert 2027: " + formatNumber(mood2027.average()))
println("Differenz:      " + formatNumber(meanDifference))
println()
println("Gepaarter t-Test (" + pairedDays.size + " Tagespaare)")
println("  t:        " + formatNumber(pairedT))
println("  p-Wert:   " + formatNumber(pairedP))
println("  Ergebnis: " + testResult(pairedP))
println()
println("Normaler t-Test (ungepaart, gleiche Varianzen)")
println("  t:        " + formatNumber(studentT))
println("  p-Wert:   " + formatNumber(studentP))
println("  Ergebnis: " + testResult(studentP))
```
