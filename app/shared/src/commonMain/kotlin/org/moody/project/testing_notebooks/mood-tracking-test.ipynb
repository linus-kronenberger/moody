## t-Tests: Mittelwertvergleich nach Jahr

H₀: Die Mittelwerte sind gleich. Alle vorhandenen Jahre werden paarweise verglichen, zweiseitig mit α = 0,05. Die p-Werte sind ohne Mehrfachtestkorrektur.


```kotlin
@file:DependsOn("org.apache.commons:commons-math3:3.6.1")

import java.time.LocalDate
import java.util.Locale
import org.apache.commons.math3.stat.inference.TTest
import org.apache.commons.math3.distribution.TDistribution
import org.apache.commons.math3.stat.StatUtils
import kotlin.math.sqrt

val moodDates = df.getColumn("Date").toList().map { LocalDate.parse(it.toString()) }
val moodValues = df.getColumn("Mood").toList().map { (it as Number).toDouble() }
require(moodDates.distinct().size == moodDates.size) { "Pro Datum darf es nur einen Mood-Wert geben." }
require(moodValues.all { it.isFinite() }) { "Mood-Werte müssen gültige Zahlen sein." }

val yearlySamples = moodDates.zip(moodValues).sortedBy { it.first }
    .groupBy { it.first.year }.toSortedMap()
val years = yearlySamples.keys.toList()
val tTests = TTest()
val significanceLevel = 0.05

data class TestInterval(val comparison: String, val test: String,
    val difference: Double, val lower: Double, val upper: Double)
val testIntervals = mutableListOf<TestInterval>()

fun addInterval(comparison: String, test: String, difference: Double, standardError: Double, df: Int) {
    val margin = TDistribution(df.toDouble()).inverseCumulativeProbability(1 - significanceLevel / 2) * standardError
    testIntervals.add(TestInterval(comparison, test, difference, difference - margin, difference + margin))
}

fun formatNumber(value: Double) = String.format(Locale.GERMANY, "%.4f", value)
fun formatPValue(p: Double) =
    if (p > 0 && p < 0.0001) String.format(Locale.GERMANY, "%.2e", p) else formatNumber(p)
fun printTest(name: String, t: Double, p: Double) {
    println(name)
    println("  t: " + formatNumber(t) + " | p-Wert: " + formatPValue(p))
    println("  Ergebnis: " + when {
        p.isNaN() -> "Nicht berechenbar (keine Streuung)"
        p < significanceLevel -> "Signifikanter Unterschied (H0 verwerfen)"
        else -> "Kein signifikanter Unterschied (H0 nicht verwerfen)"
    })
}

for ((year, samples) in yearlySamples) {
    println("Mittelwert " + year + ": " + formatNumber(samples.map { it.second }.average()) +
        " (" + samples.size + " Werte)")
}
if (years.size < 2) println("Für einen Vergleich werden mindestens zwei Jahre benötigt.")

for (i in years.indices) for (j in i + 1 until years.size) {
    val first = yearlySamples.getValue(years[i]).associate { it.first.dayOfYear to it.second }
    val second = yearlySamples.getValue(years[j]).associate { it.first.dayOfYear to it.second }
    val commonDays = first.keys.intersect(second.keys).sorted()
    val comparison = years[i].toString() + " vs. " + years[j]

    println()
    println(comparison)
    if (commonDays.size >= 2) {
        val pairedFirst = commonDays.map { first.getValue(it) }.toDoubleArray()
        val pairedSecond = commonDays.map { second.getValue(it) }.toDoubleArray()
        printTest("Gepaarter t-Test (" + commonDays.size + " gemeinsame Tage im Jahr)",
            tTests.pairedT(pairedFirst, pairedSecond), tTests.pairedTTest(pairedFirst, pairedSecond))
        val differences = pairedFirst.zip(pairedSecond) { a, b -> a - b }.toDoubleArray()
        addInterval(comparison, "Gepaart", differences.average(),
            sqrt(StatUtils.variance(differences) / differences.size), differences.size - 1)
    } else println("Gepaarter t-Test: Zu wenige gemeinsame Tage im Jahr.")

    // Der normale Student-t-Test nutzt alle Werte und setzt gleiche Varianzen voraus.
    if (first.size >= 2 && second.size >= 2) {
        val allFirst = first.values.toDoubleArray()
        val allSecond = second.values.toDoubleArray()
        printTest("Normaler t-Test (ungepaart, gleiche Varianzen)",
            tTests.homoscedasticT(allFirst, allSecond), tTests.homoscedasticTTest(allFirst, allSecond))
        val df = allFirst.size + allSecond.size - 2
        val pooledVariance = ((allFirst.size - 1) * StatUtils.variance(allFirst) +
            (allSecond.size - 1) * StatUtils.variance(allSecond)) / df
        addInterval(comparison, "Ungepaart", allFirst.average() - allSecond.average(),
            sqrt(pooledVariance * (1.0 / allFirst.size + 1.0 / allSecond.size)), df)
    } else println("Normaler t-Test: Zu wenige Werte in einem der Jahre.")
}
```

## Visualisierung der t-Tests

Die Punkte zeigen die Mittelwertdifferenz, die Balken das 95%-Konfidenzintervall. Ein Intervall vollständig über oder unter der Nulllinie entspricht einem signifikanten Unterschied bei α = 0,05. Jeder Jahresvergleich enthält beide Tests.


```kotlin
if (testIntervals.isEmpty()) {
    println("Keine Jahresvergleiche für die Grafik vorhanden.")
} else {
    val plotWidth: Int = java.lang.Math.max(900, testIntervals.size * 150)
    val intervalData = dataFrameOf(
        "Vergleich" to testIntervals.map { it.comparison + System.lineSeparator() + it.test },
        "Test" to testIntervals.map { it.test },
        "Differenz" to testIntervals.map { it.difference },
        "Untergrenze" to testIntervals.map { it.lower },
        "Obergrenze" to testIntervals.map { it.upper }
    )
    val intervalPlot = intervalData.plot {
        hLine {
            yIntercept.constant(0.0)
            color = Color.hex("#94a3b8")
            type = LineType.DASHED
        }
        errorBars {
            x("Vergleich")
            yMin("Untergrenze")
            yMax("Obergrenze")
            borderLine.color("Test")
            borderLine.width = 2.0
            width = 0.2
        }
        points {
            x("Vergleich")
            y("Differenz")
            color("Test")
            size = 5.0
        }
        layout {
            title = "Mittelwertunterschiede zwischen den Jahren"
            caption = "Punkte: Differenz (erstes − zweites Jahr) | Balken: 95%-Konfidenzintervalle | Nulllinie: gleiche Mittelwerte"
            size = plotWidth to 450
        }
    }
    DISPLAY(intervalPlot)
}
```
