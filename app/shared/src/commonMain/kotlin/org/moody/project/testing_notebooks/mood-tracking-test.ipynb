## t-Tests: Comparing Mean Mood by Year

H0: The means are equal. All available years are compared pairwise, two-sided, with alpha = 0.05. The p-values are not corrected for multiple testing.

**Limitations (please read):**
- Daily mood values are autocorrelated (today depends on yesterday). The t-test assumes independent observations, so the p-values here are overly optimistic (significance is reached too easily).
- Pairing by calendar day (same month/day across years) is a convention, not a causal relationship. There is no reason the mood on March 15th should correlate across years.
- For a year-over-year comparison, descriptive statistics and effect sizes are more robust than a significance test; treat the results below as illustrative.

```kotlin
@file:DependsOn("org.apache.commons:commons-math3:3.6.1")

import java.time.LocalDate
import java.time.MonthDay
import java.util.Locale
import org.apache.commons.math3.stat.inference.TTest
import org.apache.commons.math3.distribution.TDistribution
import org.apache.commons.math3.stat.StatUtils
import kotlin.math.sqrt

val moodDates = df.getColumn("Date").toList().map { LocalDate.parse(it.toString()) }
val moodValues = df.getColumn("Mood").toList().map { (it as Number).toDouble() }

val samplesByYear = moodDates.zip(moodValues).groupBy { it.first.year }.toSortedMap()
val years = samplesByYear.keys.toList()
val tTests = TTest()
val alpha = 0.05

data class TestInterval(val comparison: String, val test: String,
    val difference: Double, val lower: Double, val upper: Double)
val testIntervals = mutableListOf<TestInterval>()

fun format(value: Double) = String.format(Locale.US, "%.4f", value)
fun formatP(p: Double) =
    if (p in 0.0..0.0001) String.format(Locale.US, "%.2e", p) else format(p)

fun report(comparison: String, test: String, t: Double, p: Double,
    difference: Double, standardError: Double, degreesOfFreedom: Int) {
    val result = when {
        p.isNaN() -> "Not computable (no variation)"
        p < alpha -> "Significant difference (reject H0)"
        else -> "No significant difference (do not reject H0)"
    }
    println("  $test")
    println("    t: ${format(t)} | p-value: ${formatP(p)} | $result")
    val margin = TDistribution(degreesOfFreedom.toDouble()).inverseCumulativeProbability(1 - alpha / 2) * standardError
    testIntervals += TestInterval(comparison, test, difference, difference - margin, difference + margin)
}

for ((year, samples) in samplesByYear) {
    println("Mean $year: ${format(samples.map { it.second }.average())} (${samples.size} values)")
}
if (years.size < 2) println("At least two years are required for a comparison.")

for (i in years.indices) for (j in i + 1 until years.size) {
    // Match days by (month, day) so a missing leap day (e.g. 2028-02-29) does not shift the alignment.
    val first = samplesByYear.getValue(years[i]).associate { MonthDay.from(it.first) to it.second }
    val second = samplesByYear.getValue(years[j]).associate { MonthDay.from(it.first) to it.second }
    val comparison = "${years[i]} vs. ${years[j]}"
    println("\n$comparison")

    val commonDays = first.keys.intersect(second.keys)
    if (commonDays.size >= 2) {
        val a = commonDays.map { first.getValue(it) }.toDoubleArray()
        val b = commonDays.map { second.getValue(it) }.toDoubleArray()
        val diffs = a.zip(b) { x, y -> x - y }.toDoubleArray()
        report(comparison, "Paired", tTests.pairedT(a, b), tTests.pairedTTest(a, b),
            diffs.average(), sqrt(StatUtils.variance(diffs) / diffs.size), diffs.size - 1)
    } else println("  Paired: too few common calendar days.")

    // Standard Student t-test: uses all values, assumes equal variances.
    val a = first.values.toDoubleArray()
    val b = second.values.toDoubleArray()
    if (a.size >= 2 && b.size >= 2) {
        val degreesOfFreedom = a.size + b.size - 2
        val pooledVariance = ((a.size - 1) * StatUtils.variance(a) + (b.size - 1) * StatUtils.variance(b)) / degreesOfFreedom
        report(comparison, "Unpaired", tTests.homoscedasticT(a, b), tTests.homoscedasticTTest(a, b),
            a.average() - b.average(), sqrt(pooledVariance * (1.0 / a.size + 1.0 / b.size)), degreesOfFreedom)
    } else println("  Unpaired: too few values in one of the years.")
}
```

## Visualizing the t-Tests

The points show the mean difference, the bars the 95% confidence interval. An interval entirely above or below the zero line corresponds to a significant difference at alpha = 0.05. Each year comparison contains both tests.

```kotlin
if (testIntervals.isEmpty()) {
    println("No year comparisons available for the chart.")
} else {
    val plotWidth: Int = java.lang.Math.max(900, testIntervals.size * 150)
    val intervalData = dataFrameOf(
        "Comparison" to testIntervals.map { it.comparison + System.lineSeparator() + it.test },
        "Test" to testIntervals.map { it.test },
        "Difference" to testIntervals.map { it.difference },
        "LowerBound" to testIntervals.map { it.lower },
        "UpperBound" to testIntervals.map { it.upper }
    )
    val intervalPlot = intervalData.plot {
        hLine {
            yIntercept.constant(0.0)
            color = Color.hex("#94a3b8")
            type = LineType.DASHED
        }
        errorBars {
            x("Comparison")
            yMin("LowerBound")
            yMax("UpperBound")
            borderLine.color("Test")
            borderLine.width = 2.0
            width = 0.2
        }
        points {
            x("Comparison")
            y("Difference")
            color("Test")
            size = 5.0
        }
        layout {
            title = "Mean Differences Between Years"
            caption = "Points: difference (first - second year) | Bars: 95% confidence intervals | Zero line: equal means"
            size = plotWidth to 450
        }
    }
    DISPLAY(intervalPlot.autosave())
}
```

## PHQ-9: Scores, Evaluation and Histograms

The CSV contains only the nine individual answers (0-3 points each). Kotlin sums them per day to the PHQ-9 total score from 0 to 27. All available years are detected automatically.

The answers are synthetic test data that roughly match the mood of the last 14 calendar days. At the start of the data and on missing days, only the available period is used. The stage names follow the attached questionnaire phq-9-aok.pdf and are not clinical diagnoses here. No t-tests are performed for PHQ-9.

**Limitations (please read):**
- The PHQ-9 is a retrospective 2-week instrument. Here it is used as a synthetic daily value, which is conceptually inconsistent with the original instrument. A rolling 2-week window would be more faithful.
- The stage labels ("healthy", "unremarkable", etc.) are shown per day for illustration only and must not be interpreted as clinical diagnoses.

```kotlin
import java.util.Locale

val phqYears = df.getColumn("Date").toList().map { it.toString().take(4) }
val phqAnswers = (1..9).map { q -> df.getColumn("PHQ9_$q").toList().map { (it as Number).toInt() } }

// Daily total = sum of the nine answers for that day.
val phqScores = phqYears.indices.map { row -> phqAnswers.sumOf { it[row] } }
val phqScoresByYear = phqYears.zip(phqScores).groupBy({ it.first }, { it.second }).toSortedMap()

// Stages from phq-9-aok.pdf.
val phqStages = linkedMapOf(
    "0-4: healthy" to 0..4,
    "5-9: unremarkable" to 5..9,
    "10-14: mild" to 10..14,
    "15-19: moderate" to 15..19,
    "20-27: severe" to 20..27
)

fun format(value: Double, decimals: Int = 2) = String.format(Locale.US, "%.${decimals}f", value)

println("Overall PHQ-9 mean: ${format(phqScores.average())}")
for ((year, scores) in phqScoresByYear) {
    println("\nPHQ-9 $year | Mean: ${format(scores.average())} | Days: ${scores.size}")
    for ((stage, range) in phqStages) {
        val count = scores.count { it in range }
        println("  $stage: $count days (${format(100.0 * count / scores.size, 1)} %)")
    }
}
```

## PHQ-9: One-Sided Test — Did Mood Stay Equal/Better or Get Worse?

For each pair of consecutive years we test, in a directed (one-sided) way, whether mood **got worse** in the newer year. Because a *higher* PHQ-9 score means *more* depressive symptoms (worse mood), "equal or better" corresponds to a PHQ-9 score that did not increase.

- **H0 (null hypothesis):** mood in the new year is **equal or better** than the previous year, i.e. PHQ-9(new) <= PHQ-9(previous).
- **H1 (alternative):** mood **got worse**, i.e. PHQ-9(new) > PHQ-9(previous).

A t-test is not appropriate here: the daily PHQ-9 total is an ordinal, discrete, non-normal score. We therefore use a **one-sided Mann-Whitney U test** (rank-based, non-parametric). The p-value is computed from the normal approximation of U with tie correction, in the direction of H1 (new year stochastically larger than the previous year). Rejecting H0 means there is significant evidence that mood deteriorated.

**Limitations (please read):**
- Daily PHQ-9 values are autocorrelated (today depends on yesterday). The test assumes independent observations, so the p-values are overly optimistic (significance is reached too easily).
- Multiple consecutive-year comparisons are run; the p-values are **not corrected for multiple testing**.
- The rank-based test compares distributions (stochastic dominance), not the means directly.

```kotlin
import org.apache.commons.math3.distribution.NormalDistribution

val phqAlpha = 0.05

fun formatP(p: Double) =
    if (p in 0.0..0.0001) String.format(Locale.US, "%.2e", p) else format(p, 4)

// One-sided Mann-Whitney U test via the normal approximation with tie correction.
// Returns U for the "new" sample and the p-value for H1: new is stochastically LARGER
// than previous (i.e. higher PHQ-9 = worse mood). H0: new <= previous (equal or better).
data class OneSidedResult(val uNew: Double, val z: Double, val p: Double)

fun mannWhitneyOneSidedWorse(previous: DoubleArray, newYear: DoubleArray): OneSidedResult {
    val n1 = previous.size
    val n2 = newYear.size
    val combined = (previous.toList() + newYear.toList())
    val order = combined.indices.sortedBy { combined[it] }
    // Assign average ranks (1-based), handling ties.
    val ranks = DoubleArray(combined.size)
    var i = 0
    while (i < order.size) {
        var j = i
        while (j + 1 < order.size && combined[order[j + 1]] == combined[order[i]]) j++
        val avgRank = (i + 1 + j + 1) / 2.0
        for (k in i..j) ranks[order[k]] = avgRank
        i = j + 1
    }
    // Rank sum of the "new" sample (second block in `combined`).
    val rankSumNew = (n1 until combined.size).sumOf { ranks[it] }
    val uNew = rankSumNew - n2 * (n2 + 1) / 2.0
    val meanU = n1.toDouble() * n2 / 2.0
    // Tie-corrected variance of U.
    val n = n1 + n2
    val tieTerm = combined.groupingBy { it }.eachCount().values.sumOf { t ->
        val td = t.toDouble(); td * td * td - td
    }
    val varU = n1.toDouble() * n2 / 12.0 * ((n + 1) - tieTerm / (n.toDouble() * (n - 1)))
    if (varU <= 0.0) return OneSidedResult(uNew, Double.NaN, Double.NaN)
    // Continuity correction toward H1 (uNew larger than expected).
    val z = (uNew - meanU - 0.5) / Math.sqrt(varU)
    val p = 1.0 - NormalDistribution().cumulativeProbability(z) // upper tail = evidence mood got worse
    return OneSidedResult(uNew, z, p)
}

// phqScoresByYear comes from the PHQ-9 summary cell: year label -> list of daily total scores.
val phqYearsSorted = phqScoresByYear.keys.toList() // sorted map -> chronological
if (phqYearsSorted.size < 2) {
    println("At least two years are required to compare consecutive years.")
} else {
    println("One-sided Mann-Whitney U per consecutive year pair")
    println("H0: new year equal or better (PHQ-9 not increased) | H1: mood got worse (PHQ-9 increased)\n")
    for (idx in 0 until phqYearsSorted.size - 1) {
        val previousYear = phqYearsSorted[idx]
        val newYear = phqYearsSorted[idx + 1]
        val previous = phqScoresByYear.getValue(previousYear).map { it.toDouble() }.toDoubleArray()
        val current = phqScoresByYear.getValue(newYear).map { it.toDouble() }.toDoubleArray()
        val meanPrev = previous.average()
        val meanNew = current.average()
        val result = mannWhitneyOneSidedWorse(previous, current)
        val verdict = when {
            result.p.isNaN() -> "Not computable (no variation)"
            result.p < phqAlpha -> "Reject H0: significant evidence mood got WORSE"
            else -> "Do not reject H0: no significant evidence of worsening (equal or better)"
        }
        println("$previousYear -> $newYear  (mean PHQ-9 ${format(meanPrev, 2)} -> ${format(meanNew, 2)})")
        println("  U_new: ${format(result.uNew, 1)} | z: ${format(result.z, 4)} | p-value (one-sided): ${formatP(result.p)}")
        println("  $verdict\n")
    }
}
```
